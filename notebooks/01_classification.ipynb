# WeedSAM — 01: Multi-label Classification

## Project Overview

This project explores weakly supervised semantic segmentation of crops and weeds using image-level annotations.

Our approach combines:

- Multi-label classification with ResNet18.
- Grad-CAM for weak localization.
- SAM 2 for pseudo-mask refinement.
- Quantitative evaluation using IoU and Dice.

**Research question:** Can SAM 2 improve the quality of pseudo-masks generated from Grad-CAM?
